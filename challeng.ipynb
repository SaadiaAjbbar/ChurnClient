```python
#Demande à l’utilisateur de saisir la taille du tableau et ses éléments.
import pandas as pd
import numpy as np
#la taille
tab=[]
taille=int(input("entrer la taille de tableau"))
print(f"la taille est:{taille}")
#les elements
i=0
while i<taille:
   tab.append(int(input("entrer la valeur de l'element num")))
   i=i+1
#lire tableau
print(tab)

```

```python
#Demande à l’utilisateur de saisir la valeur cible target.
target=int(input("entrer la valeur cible"))
print(f"la target est:{target}")
```

```python
#Retourne les indices des deux nombres dont la somme vaut la cible.
for i in range(taille):
    for j in range(taille-1):
        if tab[i]+tab[j]==target:
            print(f"les indices somme de target sont :{i},{j} car {tab[i]} + {tab[j]} = {target}")
        
```
